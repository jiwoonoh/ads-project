```python
# something like:
# Once per dataset, not inside the loss function
# Logits shape: [N, C]
import torch

def get_all_logits_over_dataset(model, dataloader, device='cpu'):
    """
    Passes all samples in a dataloader through the model and returns
    the raw logits as a single tensor of shape [N, C].

    Args:
        model (torch.nn.Module): Trained model (not in training mode).
        dataloader (DataLoader): DataLoader for the dataset.
        device (str or torch.device): Device to run inference on.

    Returns:
        torch.Tensor: All logits with shape [N, num_classes]
    """
    model.eval()
    all_logits = []

    with torch.no_grad():
        for batch in dataloader:
            # unpack (x, y) or (x, y, ...) depending on your dataset
            if isinstance(batch, (tuple, list)):
                x = batch[0].to(device)
            else:
                x = batch.to(device)

            logits = model(x)
            all_logits.append(logits.cpu())

    return torch.cat(all_logits, dim=0)


logits_all = get_all_logits_over_dataset(model, dataloader)  # Your method
expected_logits = logits_all.mean(dim=0)  # shape [C]
```
